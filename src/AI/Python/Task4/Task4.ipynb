```python
class MultiplicationTable:
    """
    A class to print the multiplication table of a number.

    Attributes:
        number (int): The number used to create the multiplication table.
    """

    def __init__(self, number):
        self.number = number

    def print_table(self):
        """
        Prints the multiplication table from 1 to 10.
        """
        for i in range(1, 11):
            print(self.number, "*", i, "=", self.number * i)


table = MultiplicationTable(9)
table.print_table()
```

```python
class TwinPrimes:
    """
    A class to find twin prime numbers within a given limit.

    Attributes:
        limit (int): The maximum number to check.
    """

    def __init__(self, limit):
        self.limit = limit

    def is_prime(self, number):
        """
        Checks if a number is prime.

        Returns:
            bool: True if the number is prime, otherwise False.
        """
        if number < 2:
            return False

        for i in range(2, number):
            if number % i == 0:
                return False

        return True

    def print_twin_primes(self):
        """
        Prints all twin prime numbers below the limit.
        """
        for number in range(3, self.limit, 2):
            if number + 2 < self.limit:
                if self.is_prime(number) and self.is_prime(number + 2):
                    print(number, "and", number + 2)


twins = TwinPrimes(1000)
twins.print_twin_primes()
```

```python
class PrimeFactors:
    """
    A class to find the prime factors of a number.

    Attributes:
        number (int): The number to find its prime factors.
    """

    def __init__(self, number):
        self.number = number

    def get_factors(self):
        """
        Finds the prime factors of the number.

        Returns:
            list: A list containing the prime factors.
        """
        factors = []
        number = self.number
        divisor = 2

        while number > 1:
            if number % divisor == 0:
                factors.append(divisor)
                number = number // divisor
            else:
                divisor += 1

        return factors


number = PrimeFactors(56)
print(number.get_factors())
```

```python
class DecimalToBinary:
    """
    A class to convert a decimal number to binary.

    Attributes:
        number (int): The decimal number to convert.
    """

    def __init__(self, number):
        self.number = number

    def convert(self):
        """
        Converts the decimal number to binary.

        Returns:
            str: The binary representation of the number.
        """
        if self.number == 0:
            return "0"

        number = self.number
        binary = ""

        while number > 0:
            remainder = number % 2
            binary = str(remainder) + binary
            number = number // 2

        return binary


binary_number = DecimalToBinary(11)
print(binary_number.convert())
```

```python
class PerfectNumbers:
    """
    A class to find perfect numbers within a given range.

    Attributes:
        start (int): The start of the range.
        end (int): The end of the range.
    """

    def __init__(self, start, end):
        self.start = start
        self.end = end

    def is_perfect(self, number):
        """
        Checks if a number is perfect.

        Returns:
            bool: True if the number is perfect, otherwise False.
        """
        if number <= 0:
            return False

        total = 0

        for divisor in range(1, number):
            if number % divisor == 0:
                total += divisor

        return total == number

    def print_perfect_numbers(self):
        """
        Prints all perfect numbers within the given range.
        """
        for number in range(self.start, self.end + 1):
            if self.is_perfect(number):
                print(number)


perfect = PerfectNumbers(0, 100)
perfect.print_perfect_numbers()
```
