## <mark>The above result show some missing values exist in some columns So let's display them </mark>

```python
# count of missing values in each column (only columns that have nulls)
nulls = gbf.isnull().sum()
nulls[nulls > 0]
```

> ## <mark>The result above shows many missing data </mark>



###  To fix this issue I will fill these values with **median** for numeric and **mode** for categorical

- station id / name have 197 missing values only
- birth year and gender have 8,265 missing values (about 4.5% of the data) , so dropping them is not a good idea
- I used the median not the mean for birth year because it has some very old values (1878 , 1900) that affect the mean

```python
# trying to fill missing values in my datset using for loop 
# numeric -> median , text -> mode
# member_gender is skipped here, filling it with the mode (Male) will make the gender results biased

for col in gbf.columns:
    if gbf[col].isnull().sum() == 0 or col == 'member_gender':
        continue
    if pd.api.types.is_numeric_dtype(gbf[col]):
        gbf[col] = gbf[col].fillna(gbf[col].median())
    else:
        gbf[col] = gbf[col].fillna(gbf[col].mode()[0])
```

```python
#check the Nulls 
gbf.isnull().sum()
```

### **And with a proper category for other (member gender)**

```python
##  still got a missing value in the member_gender, so let's explore this column more 

print(gbf.member_gender.value_counts(dropna=False))
# print(gbf.member_gender.head(15))
```

```python
# fill  dataset with missing values.
# we already have 'Other' category so I will put the unknown gender there
gbf['member_gender'] = gbf['member_gender'].fillna('Other')
```

```python
print(gbf.member_gender.value_counts())
# print(gbf.member_gender.head(15))
```

```python
## check my result above 
gbf.isnull().sum().sum()
```

> ## <mark>The result above shows that we deal with all missing data </mark>

The unknown gender rows went to `Other` , so `Other` is now 11,917 rows instead of 3,652. I will keep that in mind when I read the gender charts.

### Now let's check duplicate value

```python
# only 4 rows are repeated , they are exact copies so I will drop them
gbf = gbf.drop_duplicates().reset_index(drop=True)
print('duplicated rows after drop:', gbf.duplicated().sum())
print(gbf.shape)
```

### Now let's check the category columns to make more cleaning dataset 

```python
### check the category columns that included object
cat_cols = gbf.select_dtypes(exclude='number').columns
for col in cat_cols:
    print(col, ':', gbf[col].nunique())

print()
print(gbf.user_type.unique())
print(gbf.member_gender.unique())
print(gbf.bike_share_for_all_trip.unique())
```

 > ## From all previous results above I found  :
 
 
 
 **=========================================================**
 
 

> 1 ) **unique Value** as in ((`member_gender`)) : Not significant , so won't clean it



> 2 ) **Incorrect data type** as in (( `user_type` &&	`member_gender`)) :  significant change , so I need to clean it.


> 3 ) **Good factor** for marketing work as in ((`member_birth_year`)) : so I need to clean it.

```python
##  step  2 ) Incorrect data type as in (( `user_type` &&	`member_gender`)) :  significant change , so  need to clean it.
# they have only 2 or 3 values so category type is better than object

for col in ['user_type', 'member_gender', 'bike_share_for_all_trip']:
    gbf[col] = gbf[col].astype('category')
```

```python
## check your work 
gbf[['user_type', 'member_gender', 'bike_share_for_all_trip']].dtypes
```

```python
## step 3 ) Good factor for marketing work as in ((`member_birth_year`)) : so need to clean it.
## To know the target audience's age and serve on them

# create a column for users' age depending on the current year
# note: trips are from Feb 2019 , so I used 2019 as the current year (age at the time of the trip)
# if we use 2025 every user will be 6 years older than the real age in the data

current_year = 2019
gbf['age'] = current_year - gbf['member_birth_year']
```

```python
## check your work 
gbf[['member_birth_year', 'age']].head()
```

```python
## change the new column age data type from float to int
gbf['age'] = gbf['age'].astype(int)
gbf['age'].dtype
```

```python
## create a function to know the average users ages
def average_age(df, col='age'):
    return round(df[col].mean(), 1)

print('Average users age :', average_age(gbf))
```

```python
## another important factor to offer  marketing services is the trip duration time 

## need to change the time from second to minute to be more usable
gbf['duration_min'] = (gbf['duration_sec'] / 60).round(2)
```

```python
## I need to change the time from second to Houre  to be more usable
gbf['duration_hour'] = (gbf['duration_sec'] / 3600).round(2)
```

```python
## check your work 
gbf[['duration_sec', 'duration_min', 'duration_hour']].head()
```

# <a id="1"> 1 ) . What is the distribution for User type?</a>

```python
## I will import seaborn library to support me in making visualizations 

import seaborn as sns

plt.figure(figsize=(8, 5))
ax = sns.countplot(data=gbf, x='user_type', color='#3a7ca5')

# write the percentage over each bar
for p in ax.patches:
    pct = p.get_height() / len(gbf) * 100
    ax.annotate(f'{pct:.1f}%', (p.get_x() + p.get_width() / 2, p.get_height()), ha='center', va='bottom')

plt.xlabel('User type')
plt.ylabel('Frequency')
plt.title('Distribution of users by user type')
plt.show()
```

> ### <mark>Your analsys and exploration here </mark>

- The majority of trips are made by **Subscribers (about 89%)**.
- **Customers** (casual users) are only about **11%** of the trips.
- So the service depends mostly on the members who have a monthly or yearly subscription.

# <a id="4">  4 ) . Do the users' ages have outlier values?</a>


```python
plt.figure(figsize=(10, 3))
sns.boxplot(data=gbf, x='age', color='#f2cc8f')
plt.title('Boxplot for users ages')
plt.xlabel('Age (years)')
plt.show()
```

```python
# IQR method
q1 = gbf['age'].quantile(0.25)
q3 = gbf['age'].quantile(0.75)
iqr = q3 - q1
upper = q3 + 1.5 * iqr
print('Upper limit (IQR) :', upper)
print('Trips above the IQR limit :', (gbf['age'] > upper).sum())
print('Trips with age more than 100 :', (gbf['age'] > 100).sum())

gbf[gbf['age'] > 100][['member_birth_year', 'age', 'user_type', 'member_gender']].head(10)
```

```python
# ages above 100 are not real (birth year 1878 , 1900 ...) so I will remove them
# ages from 55 to 90 can be real people so I will keep them
gbf = gbf[gbf['age'] <= 100].reset_index(drop=True)

print(gbf.shape)
print('max age now :', gbf['age'].max())
print('Average users age :', average_age(gbf))
```

> ### <mark>Your analsys and exploration here </mark>

- Yes , the boxplot shows many points above the upper whisker.
- Using the IQR rule the upper limit is 54.5 years , but 9,078 trips are above it and ages like 60 or 70 are normal for real people , so I didn't remove them.
- The real problem is ages **more than 100** (72 trips) , most of them have birth year 1900 and one has 1878. These are wrong entries so I removed them.
- After cleaning the max age is 99 and the average is still 34.1 , so the removed rows didn't change a lot.
