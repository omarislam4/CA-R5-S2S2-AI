**Task 1**

```python
email = "Amit_ml@gmail.edu"

if email.count("@") != 1:
    print("Invalid email")
else:
    username, remaining = email.split("@")

    if "." not in remaining:
        print("Invalid email")
    else:
        domain = remaining.rsplit(".", 1)[0]

        print("Username:", username)
        print("Domain:", domain)

        if email.endswith(".com"):
            print("Commercial Domain")
        elif email.endswith(".edu"):
            print("Educational Domain")
        else:
            print("Other Domain")
```

**Task 2**

```python
message = "###!!@mocleW EPGTQ!!!6789"

message = message.strip("#!@6789")

words = message.split()

first_word = words[0][::-1]
second_word = words[1][1:]

decoded_message = first_word + " " + second_word

print(decoded_message)
```

**Task 3**

```python
message = "&&&**$gnirtS PLIO!!@1234"

message = message.strip("&*$!@1234")

words = message.split()

first_word = words[0][::-1]

second_word = words[1]
second_word = second_word.replace("I", "E")
second_word = second_word.replace("O", "U")

decoded_message = first_word + " " + second_word

print(decoded_message)
```
